```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('collection.csv')
df.head(5)
```

```python
df.info()
df.describe()
```

```python
df.describe(include=object)
```

```python
df.isnull()
```

```python
df[df.isnull().any(axis=1)]
```

```python
df.dropna(how='any', inplace=True)
df
```

```python
sns.boxplot(y=df['gaji'])
plt.title('Boxplot Gaji Karyawan')
plt.show()
```

```python
sns.boxplot(x=df['is_merokok'])
plt.title('Boxplot Status Merokok Karyawan')
plt.show()
```

```python
sns.scatterplot(x=df['umur'], y=df['gaji'], hue=df['pekerjaan'])
plt.title('Scatterplot Usia vs Gaji Karyawan vs Pekerjaan')
plt.xlabel('Usia')
plt.ylabel('Gaji')
plt.show()
```
